# Reflection / Transmission Coefficients vs Incidence Angle

 Goals:
 - Define simple functions for SH and P–SV reflection/transmission coefficients.
 - Inputs are (rho, alpha, beta) in each layer + incidence angle in layer 1.
 - Compute complex coefficients for postcritical angles.
 - Plot amplitude (real/imag/abs) and phase (deg), similar to Fig. 6.4 for SH.

Notes:
 - We use $p = sin(\theta_1)/c1$ (Snell's law) and vertical slowness$q = \sqrt(1/c^2 - p^2)$.
 - For postcritical incidence in layer 2, $q$ becomes imaginary -> complex coefficients.
 - Phase is computed as np.angle(coef) in degrees; unwrap for cleaner curves if desired.


```python

import numpy as np
import matplotlib.pyplot as plt
```

## Utilities: ray parameter and vertical slowness

 We compute:
 - p = sin(theta1)/c1  (theta measured from vertical)
- q = sqrt(1/c^2 - p^2)  (vertical slowness)

 We work in complex arithmetic so postcritical cases are handled naturally.


```python

# %%
def ray_parameter(theta1_rad, c1):
    """Horizontal slowness p (s/km) given incidence angle theta1 and velocity c1 (km/s)."""
    return np.sin(theta1_rad) / c1

def vertical_slowness(p, c):
    """
    Vertical slowness q = sqrt(1/c^2 - p^2).
    Returns complex when p > 1/c (postcritical).
    """
    return np.sqrt((1.0 / c**2) - p**2 + 0j)

def cos_theta_from_p(p, c):
    """
    cos(theta) = c * q  since q = (1/c)*cos(theta) in real (precritical) case.
    More generally with complex q, define cos(theta) := c*q.
    """
    q = vertical_slowness(p, c)
    return c * q
```
